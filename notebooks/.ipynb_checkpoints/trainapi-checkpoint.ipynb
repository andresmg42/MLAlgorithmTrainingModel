```python
import ray
import requests
from fastapi import FastAPI
from ray import serve

app=FastAPI()

@serve.deployment
@serve.ingress(app)
class TrainAPI:
    @app.get("/")
    def root(self):
        return "Hello world"


serve.run(TrainAPI.bind(), route_prefix='/hello')
resp=request.get("http://localhost:8000/hello")
assert resp.json()=='Hello,world'
    
```
